```python
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 0) 한글 폰트 설정 (Windows)
# -----------------------------
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# -----------------------------
# 1) 1~15 항목 라벨 매핑
# -----------------------------
label_map_1_to_15 = {
    1:  "영상에 삽입되는 광고가 없다",
    2:  "제공하는 콘텐츠가 다양하다",
    3:  "최신 콘텐츠가 많다",
    4:  "오리지널 콘텐츠가 많다",
    5:  "이용요금이 적절하다",
    6:  "할인/프로모션 혜택이 크다",
    7:  "나에게 적합한 콘텐츠를 추천해준다",
    8:  "시청 가능한 기기가 다양하다",
    9:  "몰아보기로 콘텐츠를 이용하기 편리하다",
    10: "화면, 메뉴, 검색 등이 편리하다",
    11: "제공되는 영상의 화질/음질이 우수하다",
    12: "영상을 끊이지 않고 안정적으로 시청 가능하다",
    13: "제공되는 영상의 자막/더빙의 질이 우수하다",
    14: "Hollywood 콘텐츠를 시청할 수 있다",
    15: "자국 콘텐츠를 시청할 수 있다",
}

# *_*_* 형식 컬럼명을 사람이 읽기 쉬운 라벨로 바꾸는 함수
pat_three_parts = re.compile(r"^(?P<prefix>[^_]+_[^_]+)_(?P<k>\d+)$")
def pretty_label(col_name: str) -> str:
    m = pat_three_parts.match(col_name)
    if not m:
        return col_name
    k = int(m.group("k"))
    if k in label_map_1_to_15:
        # prefix는 숨기고, 항목 라벨만 사용 (축 라벨을 간결하게)
        return label_map_1_to_15[k]
    return col_name

# -----------------------------
# 2) C1_1 계열 선택 & 숫자화/이진화
# -----------------------------
c1_1_cols = [c for c in df_indonesia.columns if re.fullmatch(r"C1_1_\d+", c)]

# 숫자 변환(문자 → 숫자), NaN→0
c1_1_num = df_indonesia[c1_1_cols].apply(pd.to_numeric, errors="coerce").fillna(0)

# 0/1 이외 값이 혹시 있어도 안전하게 이진화(0은 0, 그 외는 1)
c1_1_bin = (c1_1_num != 0).astype(int)

# -----------------------------
# 3) 합계 / 비율 계산 + 라벨 매핑
# -----------------------------
counts = c1_1_bin.sum(axis=0)
ratios = c1_1_bin.mean(axis=0) * 100

# 인덱스를 사람이 읽는 한글 라벨로 교체
counts.index = [pretty_label(c) for c in counts.index]
ratios.index = [pretty_label(c) for c in ratios.index]

# 보기 좋게 정렬(내림차순)
counts = counts.sort_values(ascending=False)
ratios = ratios.sort_values(ascending=False)

# -----------------------------
# 4) 그래프: 응답 수
# -----------------------------
plt.figure(figsize=(12, 6))
plt.bar(counts.index, counts.values)
plt.title("인도네시아 - 넷플릭스 항목별 응답 수")
plt.ylabel("응답 수")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# -----------------------------
# 5) 그래프: 응답 비율(%)
# -----------------------------
plt.figure(figsize=(12, 6))
plt.bar(ratios.index, ratios.values)
plt.title("인도네시아 - 넷플릭스 항목별 응답 비율(%)")
plt.ylabel("응답 비율 (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 0) 한글 폰트 설정 (Windows)
# -----------------------------
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# -----------------------------
# 1) 1~15 항목 라벨 매핑
# -----------------------------
label_map_1_to_15 = {
    1:  "영상에 삽입되는 광고가 없다",
    2:  "제공하는 콘텐츠가 다양하다",
    3:  "최신 콘텐츠가 많다",
    4:  "오리지널 콘텐츠가 많다",
    5:  "이용요금이 적절하다",
    6:  "할인/프로모션 혜택이 크다",
    7:  "나에게 적합한 콘텐츠를 추천해준다",
    8:  "시청 가능한 기기가 다양하다",
    9:  "몰아보기로 콘텐츠를 이용하기 편리하다",
    10: "화면, 메뉴, 검색 등이 편리하다",
    11: "제공되는 영상의 화질/음질이 우수하다",
    12: "영상을 끊이지 않고 안정적으로 시청 가능하다",
    13: "제공되는 영상의 자막/더빙의 질이 우수하다",
    14: "Hollywood 콘텐츠를 시청할 수 있다",
    15: "자국 콘텐츠를 시청할 수 있다",
}

# 1~15 라벨에 대해 "항목→색" 고정 팔레트 생성 (tab20에서 15색 추출)
labels_order = [label_map_1_to_15[i] for i in range(1, 16)]
cmap = plt.cm.get_cmap('tab20', 15)
PALETTE = {lab: cmap(i) for i, lab in enumerate(labels_order)}

# *_*_* → 사람이 읽기 쉬운 라벨
pat_three_parts = re.compile(r"^(?P<prefix>[^_]+_[^_]+)_(?P<k>\d+)$")
def pretty_label(col_name: str) -> str:
    m = pat_three_parts.match(col_name)
    if not m:
        return col_name
    k = int(m.group("k"))
    return label_map_1_to_15.get(k, col_name)

# -----------------------------
# 2) 블록별 그리기 함수 (응답 수 + (참고) 비율)
# -----------------------------
# =============================
# (교체) 블록별 그리기 함수
# =============================
# -----------------------------
# 2) 블록별 그리기 함수 (응답 수 + (참고) 비율) - 제목 커스텀 지원
# -----------------------------
def plot_block_counts_and_ratios(df, block: int, service_name: str,
                                 country_name: str = "인도네시아",
                                 title_prefix: str = None,
                                 decimals: int = 1):
    # 예: block=1 -> C1_1_1 ~ C1_1_15
    col_pat = re.compile(fr"^C1_{block}_(\d+)$")
    cols = [c for c in df.columns if col_pat.fullmatch(c)]
    if not cols:
        print(f"[건너뜀] {country_name} - C1_{block}_* 컬럼이 없습니다.")
        return

    # 숫자 변환(문자→숫자). ★ NaN을 0으로 채우지 않습니다!
    num = df[cols].apply(pd.to_numeric, errors="coerce")
    valid = num.notna()                 # 응답함 (0 포함), NaN 제외
    nonzero = (num != 0) & valid        # 0이 아닌 값으로 응답

    # 응답 수(0이 아닌 값의 수)와 독립 비율(분자/분모)
    counts = nonzero.sum(axis=0)
    ratios = (counts / valid.sum(axis=0) * 100)

    # 한글 라벨로 인덱스 치환
    counts.index = [pretty_label(c) for c in counts.index]
    ratios.index = counts.index

    # 보기 좋게 정렬(응답 수 기준). 필요하면 ratios 기준으로 바꿔도 됩니다.
    counts = counts.sort_values(ascending=False)
    ratios = ratios.loc[counts.index].round(decimals)

    # 색 목록(항목 라벨별 고정색)
    colors_counts = [PALETTE.get(lbl, None) for lbl in counts.index]
    colors_ratios = [PALETTE.get(lbl, None) for lbl in ratios.index]

    # 제목 구성
    title_counts = (f"{title_prefix} - {service_name} 항목별 응답 수"
                    if title_prefix else f"{country_name} - {service_name} 항목별 응답 수")
    title_ratios = (f"{title_prefix} - {service_name} 항목별 응답 비율(%)"
                    if title_prefix else f"{country_name} - {service_name} 항목별 응답 비율(%)")

    # ---- 그래프: 응답 수 ----
    plt.figure(figsize=(12, 6))
    plt.bar(counts.index, counts.values, color=colors_counts)
    plt.title(title_counts)
    plt.ylabel("응답 수")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # ---- 그래프: 독립 비율(%) + 수치 라벨 ----
    fig, ax = plt.subplots(figsize=(12, 6))
    bars = ax.bar(ratios.index, ratios.values, color=colors_ratios)
    ax.set_title(title_ratios)
    ax.set_ylabel("응답 비율 (%)")
    ax.set_ylim(0, 100)  # 비율이므로 0~100 권장
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()

    # 막대 위 수치 라벨 (예: 37.5%)
    fmt = f"%.{decimals}f%%"
    for rect, val in zip(bars, ratios.values):
        ax.annotate(fmt % val,
                    xy=(rect.get_x() + rect.get_width() / 2, rect.get_height()),
                    xytext=(0, 3), textcoords="offset points",
                    ha="center", va="bottom")

    plt.show()



# -----------------------------
# 3) 인도네시아 블록 실행
# -----------------------------
service_map = {
    1:  "넷플릭스",
    2:  "아마존 프라임 비디오",
    3:  "디즈니+",
    40: "Viu",
    42: "Vidio",
}
```

```python
import pandas as pd

# 1) 국가 코드 매핑 (nation: 1=미국, 2=대만, 3=인도네시아)
country_map = {1: "미국", 2: "대만", 3: "인도네시아"}

# 2) D4RANK 코드 → 라벨 매핑
rank_label = {
    1: "콘텐츠 장르",
    2: "제작 국가",
    3: "알고리즘 추천",
    4: "출연진",
    5: "감독/작가",
    6: "SNS/유튜브 내 인기",
    7: "리뷰사이트 평점",
    8: "주변인 추천",
    9: "수상 내역",
    10: "태그",
}

def make_rank_table(df, nation_code: int):
    """nation_code(1/2/3)에 대해 D4RANK_1, D4RANK_2 분포를 표로 반환"""
    sub = df[df["nation"] == nation_code][["D4RANK_1", "D4RANK_2"]].copy()
    sub = sub.apply(pd.to_numeric, errors="coerce")  # 숫자로 변환, NaN 유지(유효응답 계산용)

    # 값 분포(유효응답만 집계). 인덱스 1~10 고정으로 정렬/누락 0 채우기
    idx = pd.Index(range(1, 11), name="코드")
    c1 = sub["D4RANK_1"].value_counts(dropna=True).reindex(idx, fill_value=0)
    c2 = sub["D4RANK_2"].value_counts(dropna=True).reindex(idx, fill_value=0)

    n1 = int(c1.sum())  # D4RANK_1 유효응답 수
    n2 = int(c2.sum())  # D4RANK_2 유효응답 수
    top2 = (c1 + c2).astype(int)
    n12 = n1 + n2       # Top2의 전체 표본(두 랭크 합산 기준)

    out = pd.DataFrame({
        "D4RANK_1_건수": c1.astype(int),
        "D4RANK_1_%": (c1 / n1 * 100).round(1) if n1 > 0 else 0,
        "D4RANK_2_건수": c2.astype(int),
        "D4RANK_2_%": (c2 / n2 * 100).round(1) if n2 > 0 else 0,
        "Top2_합계": top2,
        "Top2_%": (top2 / n12 * 100).round(1) if n12 > 0 else 0,
    })
    # 인덱스를 라벨로 치환
    out.index = [rank_label.get(i, i) for i in out.index]
    out.index.name = "요인"
    return out.sort_values("Top2_합계", ascending=False)

# 3) 국가별 표 만들기
table_id = make_rank_table(df, 3)  # 인도네시아
table_tw = make_rank_table(df, 2)  # 대만
table_us = make_rank_table(df, 1)  # 미국

# 4) 확인 (주피터면 display 권장)
print("\n=== 인도네시아 (D4RANK_1/2) ===")
print(table_id.to_string())
print("\n=== 대만 (D4RANK_1/2) ===")
print(table_tw.to_string())
print("\n=== 미국 (D4RANK_1/2) ===")
print(table_us.to_string())

# (선택) CSV로 저장
# table_id.to_csv("인도네시아_D4RANK_top2분포.csv", encoding="utf-8-sig")
# table_tw.to_csv("대만_D4RANK_top2분포.csv", encoding="utf-8-sig")
# table_us.to_csv("미국_D4RANK_top2분포.csv", encoding="utf-8-sig")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 한글 폰트(Windows)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 국가 코드 → 국가명
country_map = {1: "미국", 2: "대만", 3: "인도네시아"}

# D4RANK 코드 → 라벨
rank_label = {
    1: "콘텐츠 장르", 2: "제작 국가", 3: "알고리즘 추천", 4: "출연진", 5: "감독/작가",
    6: "SNS/유튜브 내 인기", 7: "리뷰사이트 평점", 8: "주변인 추천", 9: "수상 내역", 10: "태그",
}

def make_rank_table(df, nation_code: int):
    """nation_code(1/2/3)에 대해 D4RANK_1, D4RANK_2 분포 표 반환(건수/%)"""
    sub = df[df["nation"] == nation_code][["D4RANK_1", "D4RANK_2"]].copy()
    sub = sub.apply(pd.to_numeric, errors="coerce")  # 숫자화, NaN 유지

    idx = pd.Index(range(1, 11), name="코드")
    c1 = sub["D4RANK_1"].value_counts(dropna=True).reindex(idx, fill_value=0)
    c2 = sub["D4RANK_2"].value_counts(dropna=True).reindex(idx, fill_value=0)

    n1, n2 = int(c1.sum()), int(c2.sum())
    top2 = (c1 + c2).astype(int)
    n12 = n1 + n2

    out = pd.DataFrame({
        "D4RANK_1_건수": c1.astype(int),
        "D4RANK_1_%": (c1 / n1 * 100).round(1) if n1 > 0 else 0,
        "D4RANK_2_건수": c2.astype(int),
        "D4RANK_2_%": (c2 / n2 * 100).round(1) if n2 > 0 else 0,
        "Top2_합계": top2,
        "Top2_%": (top2 / n12 * 100).round(1) if n12 > 0 else 0,
    })
    out.index = [rank_label.get(i, i) for i in out.index]
    out.index.name = "요인"
    return out.sort_values("Top2_합계", ascending=False)

def plot_rank_bars(df, nation_code: int, topn: int = 10):
    """국가별 1·2순위 퍼센트(%)를 나란히 막대그래프로 표시"""
    country = country_map[nation_code]
    tbl = make_rank_table(df, nation_code)
    if topn: tbl = tbl.head(topn)

    labels = tbl.index.tolist()
    x = np.arange(len(labels))
    w = 0.38

    y1 = tbl["D4RANK_1_%"].values
    y2 = tbl["D4RANK_2_%"].values

    fig, ax = plt.subplots(figsize=(13, 6))
    b1 = ax.bar(x - w/2, y1, width=w, label="1순위")
    b2 = ax.bar(x + w/2, y2, width=w, label="2순위")
    ax.set_title(f"{country} - 시청 결정 요인 (1·2순위 %)")
    ax.set_ylabel("비율(%)")
    ax.set_xticks(x, labels, rotation=45, ha="right")
    ax.set_ylim(0, 100)
    ax.legend()
    plt.tight_layout()

    # 수치 라벨
    for rect in list(b1) + list(b2):
        v = rect.get_height()
        ax.annotate(f"{v:.1f}%", (rect.get_x()+rect.get_width()/2, v),
                    xytext=(0,3), textcoords="offset points",
                    ha="center", va="bottom")
    plt.show()

def plot_rank_top2(df, nation_code: int, topn: int = 10):
    """국가별 Top2(1+2순위) 퍼센트(%) 단일 막대그래프"""
    country = country_map[nation_code]
    tbl = make_rank_table(df, nation_code)
    if topn: tbl = tbl.head(topn)

    labels = tbl.index.tolist()
    y = tbl["Top2_%"].values

    fig, ax = plt.subplots(figsize=(13, 6))
    bars = ax.bar(labels, y)
    ax.set_title(f"{country} - 시청 결정 요인 Top2(1+2순위) 비율(%)")
    ax.set_ylabel("비율(%)")
    ax.set_ylim(0, 100)
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()

    for rect in bars:
        v = rect.get_height()
        ax.annotate(f"{v:.1f}%", (rect.get_x()+rect.get_width()/2, v),
                    xytext=(0,3), textcoords="offset points",
                    ha="center", va="bottom")
    plt.show()
```
