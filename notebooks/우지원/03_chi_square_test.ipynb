```python
import os
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

# 데이터 파일 경로 설정
folder_path = f"../../data/"
file_2023 = os.path.join(folder_path, "user&disney.csv")

# CSV 읽기
df = pd.read_csv(file_2023, encoding='utf-8-sig', low_memory=False)

print("데이터:", df.shape)

# 교차표
cross_tab = pd.crosstab(df['age'], df['pay_disney_2022'])

# 카이제곱 검정
chi2, p, dof, expected = chi2_contingency(cross_tab)

# Cramér's V 계산
n = cross_tab.sum().sum()
cramers_v = np.sqrt(chi2 / (n * (min(cross_tab.shape) - 1)))

print(f"p-value: {p}")
print(f"Cramér's V: {cramers_v}")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# 데이터 파일 경로 설정
folder_path = f"../../data/"
file_2023 = os.path.join(folder_path, "uesr&contents2022.csv")

# CSV 읽기
df = pd.read_csv(file_2023, encoding='utf-8-sig', low_memory=False)

print("2022 데이터:", df.shape)

route_cols = [col for col in df.columns if 'recog_route' in col and '2022' in col]
results = []

for col in route_cols:
    cross_tab = pd.crosstab(df['age'], df[col])
    chi2, p, _, _ = chi2_contingency(cross_tab)
    n = cross_tab.sum().sum()
    cramers_v = np.sqrt(chi2 / (n * (min(cross_tab.shape) - 1)))
    results.append((col, p, cramers_v))

result_df = pd.DataFrame(results, columns=['route', 'p_value', 'cramers_v'])
print(result_df)
```

```python
import os
import pandas as pd
from scipy.stats import chi2_contingency
from statsmodels.discrete.discrete_model import MNLogit

# 데이터 파일 경로 설정
folder_path = f"../../data/"
file_2023 = os.path.join(folder_path, "미국payment2.csv")

# CSV 읽기
df = pd.read_csv(file_2023, encoding='utf-8-sig', low_memory=False)

print("데이터:", df.shape)

# 교차표
cross_tab = pd.crosstab(df['age'], df['ott_cost_usa_2022'])

# 카이제곱 검정
chi2, p, dof, expected = chi2_contingency(cross_tab)

print(f"Chi2: {chi2}")
print(f"p-value: {p}")
print(f"자유도: {dof}")
print("기대빈도표:")
print(expected)

n = cross_tab.sum().sum()
cramers_v = np.sqrt(chi2 / (n * (min(cross_tab.shape)-1)))
print(f"Cramér’s V: {cramers_v}")

# age 더미 변수 변환
df_dummy = pd.get_dummies(df, columns=['age'], drop_first=True)

# 다항 로지스틱 회귀
model = MNLogit(df['ott_cost_usa_2022'], df_dummy.drop(columns=['ott_cost_usa_2022']))
result = model.fit()
print(result.summary())
```
